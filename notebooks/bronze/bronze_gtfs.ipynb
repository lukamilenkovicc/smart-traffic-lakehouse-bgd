```python
gtfs_files = {
    "stops":"stops.txt",
    "routes":"routes.txt",
    "trips":"trips.txt",
    "stop_times":"stop_times.txt",
    "shapes":"shapes.txt",
    "calendar":"calendar.txt",
    "calendar_dates":"calendar_dates.txt",
    "trips":"trips.txt"
}

source_path = "/Volumes/bg_traffic/bg_traffic_bronze/landing/gtfs"
folders = [f.path for f in dbutils.fs.ls(source_path) if f.isDir() and not f.name.startswith("_")]

latest_folder_path = sorted(folders)[-1].rstrip("/")
for name, filename in gtfs_files.items():
    file_path = f"{latest_folder_path}/{filename}"
    df = (spark.read.csv(file_path, header=True, inferSchema=True))
    target_table = f"bg_traffic.bg_traffic_bronze.gtfs_{name}"
    df.write.format("delta").mode("overwrite").saveAsTable(target_table)
    print(f"Ucitano: gtfs_{name}")

```
