### OpenMeteo


```python
%sql
CREATE TABLE IF NOT EXISTS bg_traffic.bg_traffic_silver.openmeteo(
    weather_timestamp TIMESTAMP NOT NULL,
    temperature_c DOUBLE,
    relative_humidity DOUBLE, 
    precipitation_mm DOUBLE,
    wind_speed_kmh DOUBLE,
    rain_mm DOUBLE,
    fetched_at TIMESTAMP,
    _ingestion_timestamp TIMESTAMP, 
    _source_file STRING

)
USING DELTA;
```

### OpenAQ Measurements

```python
%sql
CREATE TABLE IF NOT EXISTS bg_traffic.bg_traffic_silver.openaq_measurements(
    location_id BIGINT,
    sensor_id BIGINT,
    measurement_timestamp TIMESTAMP,
    measurement_value DOUBLE,
    fetched_at TIMESTAMP,
    _ingestion_timestamp TIMESTAMP,
    _source_file STRING
)
USING DELTA;
```

### OpenAQ Locations

```python
%sql
CREATE TABLE IF NOT EXISTS bg_traffic.bg_traffic_silver.openaq_locations(
    location_id BIGINT,
    location_name STRING,
    latitude DOUBLE,
    longitude DOUBLE,
    provider_id BIGINT,
    provider_name STRING,
    fetched_at TIMESTAMP,
    _ingestion_timestamp TIMESTAMP,
    _source_file STRING

)
USING DELTA;
```
